# Multimodal Teacher Guided DINOv3 Adapter

Small-scale validation on `zh0416/pv-gf2`: data audit, InternVL semantic cache, frozen DINOv3 SAT feature cache, and controlled A/B segmentation experiments.

```python
import json, os, shutil, subprocess, sys
from pathlib import Path
import torch

assert torch.cuda.is_available(), 'Enable a GPU accelerator in Kaggle before running.'
print('GPU:', torch.cuda.get_device_name(0))
print('CUDA:', torch.version.cuda)

try:
    from kaggle_secrets import UserSecretsClient
    token = UserSecretsClient().get_secret('HF_TOKEN')
    if token:
        os.environ['HF_TOKEN'] = token
        print('HF_TOKEN loaded from Kaggle Secrets')
except Exception:
    print('HF_TOKEN secret not available; public or mounted model paths will be used')

os.environ['HF_HOME'] = '/kaggle/temp/model_cache'
os.environ['HUGGINGFACE_HUB_CACHE'] = '/kaggle/temp/model_cache'
os.environ['TRANSFORMERS_CACHE'] = '/kaggle/temp/model_cache'

dino_weights = [p for p in Path('/kaggle/input').rglob('model.safetensors') if 'timmvit-large-patch16-dinov3-sat493m' in p.as_posix()]
assert len(dino_weights) == 1, f'Expected one mounted DINOv3 SAT checkpoint, found: {dino_weights}'
os.environ['DINOV3_MODEL_PATH'] = str(dino_weights[0])
print('DINOv3 cloud checkpoint:', dino_weights[0])
```

```python
repo = Path('/kaggle/temp/DINOV3')
if not repo.exists():
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/Zh0416/DINOV3.git', str(repo)], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', str(repo / 'requirements.txt')], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-e', str(repo), '--no-deps', '--no-build-isolation'], check=True)
print('Code commit:', subprocess.check_output(['git', '-C', str(repo), 'rev-parse', 'HEAD'], text=True).strip())
config = str(repo / 'configs' / 'kaggle_small.yaml')
```

## Stage 1 - Dataset inspection and fixed stratified split

```python
subprocess.run([sys.executable, str(repo / 'scripts' / '01_inspect_data.py'), '--config', config], check=True)

output_root = Path('/kaggle/working/outputs')
teacher_caches = [p for p in Path('/kaggle/input').rglob('semantic_prompt.json') if 'dinov3-pv-internvl-teacher-cache' in p.as_posix()]
if teacher_caches:
    cached = json.loads(teacher_caches[0].read_text(encoding='utf-8'))
    current = json.loads((output_root / 'split_manifest.json').read_text(encoding='utf-8'))
    cached_ids = {record['sample_id'] for record in cached['records']}
    current_ids = {record['sample_id'] for record in current['samples']}
    assert cached_ids == current_ids, 'Mounted teacher cache does not match the current split'
    shutil.copy2(teacher_caches[0], output_root / 'semantic_prompt.json')
    print('Restored 300-record InternVL teacher cache:', teacher_caches[0])
```
